```python
# ============================
# NLP Practical using NLTK & spaCy
# ============================

# Install (Run once in Google Colab)
!pip install -q spacy
!python -m spacy download en_core_web_sm

# Import Libraries
import nltk
import spacy

from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download Required NLTK Data
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger_eng')

# Load spaCy Model
nlp = spacy.load("en_core_web_sm")

# =====================================
# User Input Text
# =====================================

text = input("Enter a paragraph for NLP analysis:\n")

print("\nOriginal Text:\n")
print(text)

# =====================================
# Step 1: Sentence Segmentation
# =====================================

sentences = sent_tokenize(text)

print("\nSentence Segmentation:")

for i, sent in enumerate(sentences, 1):
    print(f"{i}: {sent}")

# =====================================
# Step 2: Word Tokenization
# =====================================

words = word_tokenize(text)

print("\nWord Tokenization:")
print(words)

# =====================================
# Step 3: Stop Word Removal
# =====================================

stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word.lower() not in stop_words
]

print("\nAfter Stop Word Removal:")
print(filtered_words)

# =====================================
# Step 4: Stemming
# =====================================

stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in filtered_words
]

print("\nAfter Stemming:")
print(stemmed_words)

# =====================================
# Step 5: Lemmatization
# =====================================

lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in filtered_words
]

print("\nAfter Lemmatization:")
print(lemmatized_words)

# =====================================
# Step 6: POS Tagging
# =====================================

pos_tags = nltk.pos_tag(words)

print("\nPOS Tagging:")
print(pos_tags)

# =====================================
# Step 7: Named Entity Recognition (NER)
# =====================================

doc = nlp(text)

print("\nNamed Entity Recognition (NER):")

for ent in doc.ents:
    print(ent.text, "->", ent.label_)

# =====================================
# Step 8: Dependency Parsing
# =====================================

print("\nDependency Parsing:")

for token in doc:
    print(token.text, "->", token.dep_, "->", token.head.text)

# =====================================
# Step 9: Chunking (Noun Phrases)
# =====================================

print("\nNoun Phrase Chunking:")

for chunk in doc.noun_chunks:
    print(chunk.text)
```
